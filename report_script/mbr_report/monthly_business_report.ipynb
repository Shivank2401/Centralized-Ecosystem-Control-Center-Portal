# SKU Bases

## Analysis

### create organic % current vs previous

```python
# Convert year_month to datetime
import pandas as pd

# Example:
# df columns -> ['year_month', 'sku', 'total_sales']
df = df_merged.copy()
# Convert year_month to datetime
df['year_month'] = pd.to_datetime(df['year_month'], format='%Y-%m')
# -------------------------------
# 1. Current Month % Contribution
# -------------------------------

# Total sales of each month
monthly_total_organic = df.groupby('year_month')['organic_sales'].transform('sum')

# % contribution of SKU in that month
df['pct_organic_current_month'] = (
    df['organic_sales'] / monthly_total_organic
) * 100

# -----------------------------------
# 2. Previous Month % Contribution
# -----------------------------------

# Sort properly
df = df.sort_values(['sku', 'year_month'])

# Get previous month contribution for same SKU
df['pct_organic_previous_month'] = (
    df.groupby('sku')['pct_organic_current_month']
      .shift(1)
)
df
```

### creating contribution %

```python
# Convert year_month to datetime
import pandas as pd

# Example:
# df columns -> ['year_month', 'sku', 'total_sales']

# Convert year_month to datetime
df['year_month'] = pd.to_datetime(df['year_month'], format='%Y-%m')
# -------------------------------
# 1. Current Month % Contribution
# -------------------------------

# Total sales of each month
monthly_total = df.groupby('year_month')['total_sales'].transform('sum')

# % contribution of SKU in that month
df['pct_contribution_current_month'] = (
    df['total_sales'] / monthly_total
) * 100

# -----------------------------------
# 2. Previous Month % Contribution
# -----------------------------------

# Sort properly
df = df.sort_values(['sku', 'year_month'])

# Get previous month contribution for same SKU
df['pct_contribution_previous_month'] = (
    df.groupby('sku')['pct_contribution_current_month']
      .shift(1)
)
df
```

### SKU type

```python
df['sku_type'] = np.where(df['pct_contribution_current_month']>=5, "Critical", 
                                 np.where((df['pct_contribution_current_month']>=2) & (df['pct_contribution_current_month']<5
                                                                                              ), "Non-Critical","Review"))
df
```

```python
df['sales_alert'] = np.where((df['pct_contribution_current_month'] < df['pct_contribution_previous_month']
                                     ) & (df['pct_contribution_current_month'] >= 2), "Sales Contribution Down", "Stable Sales")
df['organic_alert'] = np.where((df['pct_organic_current_month'] < df['pct_organic_previous_month']
                                       ), "Organic Contribution Down", "Stable Organic")
df
```

### Inventory Utilization %

```python
df['sellable'] = df['afn_fulfillable_quantity'] + df['fc_transfer']

df['sellable_with_inbound'] = df['sellable']  + df['afn_inbound_receiving_quantity'] 

df['inventory_utilization_pct'] = (df['quantity'] / df['sellable']) * 100
```

### `inventory_age_risk` classification

```python
# Aged inventory is defined as inventory that has been in stock for more than 180 days.
df["aged_inventory_units"] = (
    df["inv_age_181_to_270_days"] +
    df["inv_age_271_to_365_days"] +
    df["inv_age_365_plus_days"]
)
df["aged_inventory_pct"] = (df["aged_inventory_units"] / df["sellable"]) * 100


# Fresh inventory is defined as inventory that has been in stock for 90 days or less.
df["fresh_inventory"] = (
    df["inv_age_0_to_30_days"] +
    df["inv_age_31_to_60_days"] +
    df["inv_age_61_to_90_days"]
)
df["fresh_inventory_pct"] = (df["fresh_inventory"] / df["sellable"]) * 100

# Inventory age risk categorization
df["inventory_age_risk"] = np.select(
    [
        df["aged_inventory_pct"] > 30,
        df["aged_inventory_pct"].between(10, 30, inclusive="right")
    ],
    [
        "High Risk",
        "Medium Risk"
    ],
    default="Healthy"
)
```

### Month-over-Month Growth %

```python
df = df.sort_values(['sku','year_month'])

df['prev_sales'] = df.groupby('sku')['total_sales'].shift(1)

df['mom_growth_pct'] = (
    (df['total_sales'] - df['prev_sales'])
    / df['prev_sales']
) * 100
df
```

### MOM Fast-Growing SKU Detection

```python
df['growth_status_mom'] = np.where(
    df['mom_growth_pct'] > 20,
    'Fast Growing',
    
    np.where(
        df['mom_growth_pct'] < -20,
        'Declining',
        'Stable'
    )
)
```

### YOY comparison

```python
df = df.sort_values(['sku', 'year_month'])

# Previous year's same month sales
df['prev_year_sales'] = df.groupby('sku')['total_sales'].shift(12)

# YOY Growth %
df['yoy_growth_pct'] = (
    (df['total_sales'] - df['prev_year_sales'])
    / df['prev_year_sales']
) * 100
df
```

### YOY Fast-Growing SKU Detection

```python
df['growth_status_yoy'] = np.where(
    df['yoy_growth_pct'] > 20,
    'Fast Growing',
    
    np.where(
        df['yoy_growth_pct'] < -20,
        'Declining',
        'Stable'
    )
)
df
```

###  Ad Dependency Score

```python
df['ad_dependency_pct'] = (
    df['ad_sales'] / df['total_sales']
) * 100


df['ad_dependency_type'] = np.where(
    df['ad_dependency_pct'] > 70,
    'Highly Ad Dependent',

    np.where(
        (df['ad_dependency_pct'] >= 40) & 
        (df['ad_dependency_pct'] <= 70),
        'Moderate',
        'Strong Organic SKU'
    )
)

df
```

### Highly Ad Dependent + Declining Growth

```python
conditions = [

    # Condition 1
    (df['ad_dependency_pct'] > 50) & 
    (df['mom_growth_pct'] > 0) & 
    (df['mom_growth_pct'] <= 50),

    (df['ad_dependency_pct'] > 50) & 
    (df['mom_growth_pct'] > 50),

    (df['ad_dependency_pct'] > 50) & 
    (df['mom_growth_pct'] < 0),

    # Condition 2
    (df['ad_dependency_pct'] < 50) & 
    (df['mom_growth_pct'] > 0) & 
    (df['mom_growth_pct'] <= 50),

    (df['ad_dependency_pct'] < 50) & 
    (df['mom_growth_pct'] > 50),

    (df['ad_dependency_pct'] < 0) & 
    (df['mom_growth_pct'] < 0)

]

choices = [

    'Ad Driven Growth',
    'High Growth via Ads',
    'Wasted Ad Spend',

    'Healthy Organic Growth',
    'Organic Star SKU',
    'Organic Decline'

]

df['growth_ad_alert'] = np.select(
    conditions,
    choices,
    default='Stable'
)
```

### spend_growth_alert

```python
conditions = [

    # ---------------------------------------------------
    # High spend + declining sales + declining organic
    # ---------------------------------------------------
    (df['ad_spend'] > df['ad_spend'].median()) &
    (df['mom_growth_pct'] < 0) &
    (df['pct_organic_current_month'] < 0),

    # ---------------------------------------------------
    # High spend + low sales growth + declining organic
    # ---------------------------------------------------
    (df['ad_spend'] > df['ad_spend'].median()) &
    (df['mom_growth_pct'] >= 0) &
    (df['mom_growth_pct'] < 10) &
    (df['pct_organic_current_month'] < 0),

    # ---------------------------------------------------
    # High spend + strong sales growth + strong organic
    # ---------------------------------------------------
    (df['ad_spend'] > df['ad_spend'].median()) &
    (df['mom_growth_pct'] > 20) &
    (df['pct_organic_current_month'] > 10),

    # ---------------------------------------------------
    # Low spend + high sales growth + strong organic
    # ---------------------------------------------------
    (df['ad_spend'] < df['ad_spend'].median()) &
    (df['mom_growth_pct'] > 30) &
    (df['pct_organic_current_month'] > 20),

    # ---------------------------------------------------
    # Low spend + positive growth but weak organic
    # ---------------------------------------------------
    (df['ad_spend'] < df['ad_spend'].median()) &
    (df['mom_growth_pct'] > 0) &
    (df['pct_organic_current_month'] < 0)

]

choices = [

    'Severe Wasted Spend',
    'Inefficient Ad Spend',
    'Efficient Ad Scaling',
    'Organic Strong Growth',
    'Weak Organic Momentum'

]

df['spend_performance_alert'] = np.select(
    conditions,
    choices,
    default='Healthy'
)
```

### overspending_alert

```python
df['overspending_alert'] = np.where(
    (df['tacos'] > 20) &
    (df['pct_contribution_current_month'] < 2),

    'Overspending on Low Contributor',
    'Normal'
)
```

### Month-over-Month Spend Change

```python
df['prev_ad_spend'] = (
    df.groupby('sku')['ad_spend']
      .shift(1)
)

df['ad_spend_growth_pct'] = (
    (df['ad_spend'] - df['prev_ad_spend'])
    / df['prev_ad_spend']
) * 100

                         
```

### Organic Cannibalization Alert

```python
df = df.sort_values(['sku', 'year_month'])

df['roas_current_month'] = np.where(
    df['ad_spend'] > 0,
    df['ad_sales'] / df['ad_spend'],
    np.nan
)
# TACOS
df['tacos_current_month'] = df['tacos']

df['tacos_previous_month'] = (
    df.groupby('sku')['tacos']
      .shift(1)
)

# ROAS
df['roas_current_month'] = np.where(
    df['ad_spend'] > 0,
    df['ad_sales'] / df['ad_spend'],
    np.nan
)

df['roas_previous_month'] = (
    df.groupby('sku')['roas_current_month']
      .shift(1)
)
```

```python
conditions = [

    # Critical Cannibalization
    (
        (df['ad_spend_growth_pct'] > 40) &
        (df['pct_organic_current_month'] < df['pct_organic_previous_month']) &
        (df['mom_growth_pct'] < 5) &
        (df['tacos_current_month'] > df['tacos_previous_month']) &
        (df['roas_current_month'] < df['roas_previous_month'])
    ),

    # Moderate Cannibalization
    (
        (df['ad_spend_growth_pct'] > 20) &
        (df['pct_organic_current_month'] < df['pct_organic_previous_month']) &
        (df['mom_growth_pct'] < 10)
    )
]

choices = [
    'Critical Cannibalization',
    'Moderate Cannibalization'
]

df['organic_cannibalization_alert'] = np.select(
    conditions,
    choices,
    default='Healthy'
)
```

### 2. Advertising & organic sales metrics

```python
df_material["organic_sales"] = df_material["total_sales"] - df_material["ad_sales"]
df_material['organics_sales_pct'] = (df_material['organic_sales'] / df_material['total_sales']) * 100
df_material["acos"] = (df_material["ad_spend"] / df_material["ad_sales"]) * 100
df_material['tacos'] = (df_material["ad_spend"] / df_material["total_sales"]) * 100
```

### 3. Portfolio share — organic % (current vs previous)

```python
monthly_total_organic = df_material.groupby('year_month')['organic_sales'].transform('sum')
df_material['pct_organic_current_month'] = (df_material['organic_sales'] / monthly_total_organic) * 100

# Shift within each Material (was grouped by 'sku')
df_material['pct_organic_previous_month'] = df_material.groupby('Material')['pct_organic_current_month'].shift(1)
```

### 4. Portfolio share — sales contribution % (current vs previous)

```python
monthly_total = df_material.groupby('year_month')['total_sales'].transform('sum')
df_material['pct_contribution_current_month'] = (df_material['total_sales'] / monthly_total) * 100

df_material['pct_contribution_previous_month'] = df_material.groupby('Material')['pct_contribution_current_month'].shift(1)
```

### 5. Material classification

```python
df_material['material_type'] = np.where(
    df_material['pct_contribution_current_month'] >= 5, "Critical",
    np.where(
        (df_material['pct_contribution_current_month'] >= 2) &
        (df_material['pct_contribution_current_month'] < 5),
        "Non-Critical", "Review"
    )
)
```

### 6. Performance alerts (contribution & organic)

```python
df_material['sales_alert'] = np.where(
    (df_material['pct_contribution_current_month'] < df_material['pct_contribution_previous_month']) &
    (df_material['pct_contribution_current_month'] >= 2),
    "Alert", "No Alert"
)

df_material['organic_alert'] = np.where(
    df_material['pct_organic_current_month'] < df_material['pct_organic_previous_month'],
    "Alert", "No Alert"
)
```

### 7. Inventorv metrics

```python
df_material['sellable'] = df_material['afn_fulfillable_quantity'] + df_material['fc_transfer']
df_material['sellable_with_inbound'] = df_material['sellable'] + df_material['afn_inbound_receiving_quantity']
df_material['inventory_utilization_pct'] = (df_material['quantity'] / df_material['sellable']) * 100
```

### 8. Month-over-month (MoM) growth

```python
df_material['prev_sales'] = df_material.groupby('Material')['total_sales'].shift(1)
df_material['mom_growth_pct'] = ((df_material['total_sales'] - df_material['prev_sales']) / df_material['prev_sales']) * 100

df_material['growth_status_mom'] = np.where(
    df_material['mom_growth_pct'] > 20, "Fast Growing",
    np.where(df_material['mom_growth_pct'] < -20, "Declining", "Stable")
)
```

### 9. Year-over-year (YoY) growth

```python
df_material['prev_year_sales'] = df_material.groupby('Material')['total_sales'].shift(12)
df_material['yoy_growth_pct'] = ((df_material['total_sales'] - df_material['prev_year_sales']) / df_material['prev_year_sales']) * 100

df_material['growth_status_yoy'] = np.where(
    df_material['yoy_growth_pct'] > 20, "Fast Growing",
    np.where(df_material['yoy_growth_pct'] < -20, "Declining", "Stable")
)
```

### 10. Ad dependency score

```python
df_material['ad_dependency_pct'] = (df_material['ad_sales'] / df_material['total_sales']) * 100

df_material['ad_dependency_type'] = np.where(
    df_material['ad_dependency_pct'] > 70, "Highly Ad Dependent",
    np.where(
        (df_material['ad_dependency_pct'] >= 40) & (df_material['ad_dependency_pct'] <= 70),
        "Moderate", "Strong Organic SKU"
    )
)
```

### 11. Growth x ad dependency alert

```python
conditions = [
    (df_material['ad_dependency_pct'] > 50) & (df_material['mom_growth_pct'] > 0) & (df_material['mom_growth_pct'] <= 50),
    (df_material['ad_dependency_pct'] > 50) & (df_material['mom_growth_pct'] > 50),
    (df_material['ad_dependency_pct'] > 50) & (df_material['mom_growth_pct'] < 0),
    (df_material['ad_dependency_pct'] < 50) & (df_material['mom_growth_pct'] > 0) & (df_material['mom_growth_pct'] <= 50),
    (df_material['ad_dependency_pct'] < 50) & (df_material['mom_growth_pct'] > 50),
    (df_material['ad_dependency_pct'] < 50) & (df_material['mom_growth_pct'] < 0),  # ⚠️ likely intended < 50
]
choices = [
    "Ad Driven Growth", "High Growth via Ads", "Wasted Ad Spend",
    "Healthy Organic Growth", "Organic Star Material", "Organic Decline"
]
df_material['growth_ad_alert'] = np.select(conditions, choices, default="Stable")
```

### 12. Spend performance alert

```python
median_ad_spend = df_material['ad_spend'].median()  # dataset-wide median

conditions_sp = [
    (df_material['ad_spend'] > median_ad_spend) & (df_material['mom_growth_pct'] < 0),
    (df_material['ad_spend'] > median_ad_spend) & (df_material['mom_growth_pct'] >= 0) & (df_material['mom_growth_pct'] < 10),
    (df_material['ad_spend'] < median_ad_spend) & (df_material['mom_growth_pct'] > 30),
]
choices_sp = ["Wasted Ad Spend", "High Spend Low Growth", "Organic Strong Growth"]
df_material['spend_performance_alert'] = np.select(conditions_sp, choices_sp, default="Healthy")
```

### 13. Overspending alert

```python
df_material['overspending_alert'] = np.where(
    (df_material['tacos'] > 20) & (df_material['pct_contribution_current_month'] < 2),
    "Overspending on Low Contributor", "Normal"
)
```

### 14. Ad spend MoM change

```python
df_material['prev_ad_spend'] = df_material.groupby('Material')['ad_spend'].shift(1)
df_material['ad_spend_growth_pct'] = ((df_material['ad_spend'] - df_material['prev_ad_spend']) / df_material['prev_ad_spend']) * 100
```

### 15. Organic cannibalization alert

```python
df_material['organic_cannibalization_alert'] = np.where(
    (df_material['ad_spend_growth_pct'] > 20) & (df_material['pct_organic_current_month'] < 0),
    "Organic Cannibalization", "Healthy"
)
```
