```python
# load only latest date inventory data for knc uk and ie which we processed in inventory_shopify.py(shopify report )
inventory_data = pd.read_csv(r"C:\Users\Shivank Rajput\OneDrive - EcoSoul Home\Central Repository\Data_nd_Developement_Pipeline\sanity_data\KNC\manage_fba\UK_inventory_shopify.csv")
inventory_data
```

```python
# load current month and previous month ledger data for knc uk and ie which we processed in ledger_shopify.py(shopify report )
ledger_data = pd.read_csv(r"C:\Users\Shivank Rajput\OneDrive - EcoSoul Home\Central Repository\Data_nd_Developement_Pipeline\sanity_data\KNC\ledger\UK_shopify_ledger_inventory.csv")


ledger_data['ending_balance'] = pd.to_numeric(ledger_data['ending_balance'], errors='coerce').fillna(0)

# Flag In-stock = 1 if balance > 0
ledger_data['instock'] = (ledger_data['ending_balance'] > 0).astype(int)

# # Group metrics
ledger_summary = ledger_data.groupby(['year_month', 'sku']).agg(
    days_in_stock=('instock', 'sum'), total_days=('date', 'nunique') ).reset_index()

ledger_summary['days_out_of_stock'] = ledger_summary['total_days'] - ledger_summary['days_in_stock']
ledger_summary['in_stock_rate_percent'] = ((ledger_summary['days_in_stock'] / ledger_summary['total_days']) * 100).round(0)
ledger_data = ledger_summary[['year_month', 'sku','total_days' ,'days_in_stock', 'days_out_of_stock', 'in_stock_rate_percent']]
ledger_data
```

```python
today = pd.to_datetime("today")
if today.day <= 3:
    dropper = (today - pd.DateOffset(months=2)).strftime("%Y-%m")
else:
    dropper = (today - pd.DateOffset(months=1)).strftime("%Y-%m")

print(dropper)

previous_month_ledger_shopify = ledger_data[ledger_data['year_month'] == dropper]
previous_month_ledger_shopify
```

```python
# from orders report extracting previous month orders data
previous_month_orders_shopify = shopify_orders[shopify_orders['year_month'] == dropper]
previous_month_orders_shopify = pd.pivot_table(previous_month_orders_shopify, index=['year_month','sku'],
                                values='quantity', aggfunc='sum').reset_index()
previous_month_orders_shopify
```

```python
# merge previous month ledger and previous month orders data
instock_days_previous = pd.merge(previous_month_ledger_shopify, previous_month_orders_shopify, on=['year_month','sku'], how='left')

# Convert to numeric before calculations
instock_days_previous['quantity'] = pd.to_numeric(instock_days_previous['quantity'], errors='coerce').fillna(0)
instock_days_previous['days_in_stock'] = pd.to_numeric(instock_days_previous['days_in_stock'], errors='coerce').fillna(0)

instock_days_previous['avg_unit_sold_qty'] = (instock_days_previous['quantity'] / instock_days_previous['days_in_stock'].replace(0, 1)).round(0)
instock_days_previous['avg_unit_sold_qty'] = np.where(instock_days_previous['avg_unit_sold_qty'] == 0,1 , instock_days_previous['avg_unit_sold_qty']) # type: ignore
instock_days_previous = instock_days_previous.fillna(0)

instock_days_previous['mtq_overstock'] = instock_days_previous['avg_unit_sold_qty'] * 90
instock_days_previous['mtq_understock'] = instock_days_previous['avg_unit_sold_qty'] * 30
instock_days_previous = instock_days_previous.rename(
        columns={
            "days_in_stock": "previous_dis",
            "total_days": "previous_total_days",
            "days_out_of_stock": "previous_days_out_of_stock",
            "avg_unit_sold_qty": "previous_avg_unit_sold_qty",
            "instock_rate": "last_month_instock",
            "quantity": "previous_month_sold_quantity",
        }
    )
instock_days_previous
```
